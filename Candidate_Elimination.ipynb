```python
import csv

# Read the training data from CSV file
data = []

with open('enjoysport2.csv', 'r') as csvfile:
    reader = csv.reader(csvfile)
    header = next(reader)  # Skip header

    for row in reader:
        data.append(row)

print("Training Data:")
for row in data:
    print(row)

# Number of attributes
num_attributes = len(data[0]) - 1

# Initialize Specific Hypothesis
S = ['0'] * num_attributes

# Initialize General Hypothesis
G = [['?'] * num_attributes]

print("\nInitial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
print(G)


# Function to check whether hypothesis matches an example
def covers(hypothesis, example):
    for i in range(num_attributes):
        if hypothesis[i] != '?' and hypothesis[i] != example[i]:
            return False
    return True


# Candidate Elimination Algorithm
for example in data:

    attributes = example[:-1]
    label = example[-1]

    print("\n----------------------------------------")
    print("Training Example:", example)

    # If positive example
    if label == "Yes":

        # Remove general hypotheses that do not cover positive example
        G = [g for g in G if covers(g, attributes)]

        # Generalize S to cover the positive example
        for i in range(num_attributes):

            if S[i] == '0':
                S[i] = attributes[i]

            elif S[i] != attributes[i]:
                S[i] = '?'

        print("Positive Example")

    # If negative example
    else:

        print("Negative Example")

        # Remove specific hypothesis if it covers negative example
        if covers(S, attributes):
            for i in range(num_attributes):

                if S[i] != '?':
                    new_hypothesis = S.copy()
                    new_hypothesis[i] = '?'

                    # Add to G if not already present
                    if new_hypothesis not in G:
                        G.append(new_hypothesis)

        # Remove general hypotheses that cover negative example
        G = [g for g in G if not covers(g, attributes)]

    print("S =", S)
    print("G =", G)


print("\n========================================")
print("FINAL VERSION SPACE")
print("========================================")

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)
```
